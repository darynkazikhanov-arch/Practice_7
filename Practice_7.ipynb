```python
import numpy as np

b = np.array([0.50, 0.30, 0.20])
T = np.array([[0.70, 0.30, 0.00],
              [0.20, 0.60, 0.20],
              [0.00, 0.30, 0.70]])

# A: [0.8, 0.2], B: [0.5, 0.5], C: [0.2, 0.8]
sensor_model = {'RED': np.array([0.80, 0.50, 0.20]),
                'BLUE': np.array([0.20, 0.50, 0.80])}

observations = ['RED', 'BLUE', 'BLUE', 'RED']

def predict(b, T):
    return np.dot(b, T)

def update(b_pred, observation, sensor_model):
    likelihood = sensor_model[observation]
    b_upd = b_pred * likelihood
    return b_upd / np.sum(b_upd)
current_b = b
for i, obs in enumerate(observations):
    current_b = predict(current_b, T)
    current_b = update(current_b, obs, sensor_model)
    print(f"{i+1}-ші бақылау ({obs}) кейінгі ықтималдықтар (A, B, C):", np.round(current_b, 4))
    print(("Ең ықтимал аймақ: A" if np.argmax(current_b)==0 else ("Ең ықтимал аймақ: B" if np.argmax(current_b)==1 else "Ең ықтимал аймақ: C")), "\n")
```
